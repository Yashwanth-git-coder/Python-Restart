**Question 5:** Write a Python program to calculate the factorial of a number.

```python
# Your code here
num = int(input("Enter the Factorial number: "))

if num < 0:
    print("Factorial is not defined for negative numbers.")
elif num == 0:
    print("The factorial of 0 is 1.")
else:
    factorial = 1
    for i in range(1, num + 1):
        factorial *= i
    print(f"The factorial of {num} is: {factorial}")
```

## 3. Basic Operators (Arithmetic, Comparison, Logical)

**Question 11:** Write a Python program to perform arithmetic operations: addition, subtraction, multiplication, and division.

```python
# Your code here
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

operation = input("Enter the operation (+, -, *, /): ")

if operation == '+':
    result = num1 + num2
elif operation == '-':
    result = num1 - num2
elif operation == '*':
    result = num1 * num2
elif operation == '/':
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Error! Division by zero."
else:
    result = "Invalid operation."   

print(f"The result is: {result}")
```

**Question 12:** Write a Python program to demonstrate comparison operators: equal to, not equal to, greater than, less than.

```python
# Your code here
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

if num1 > num2:
    print(f"{num1} is greater than {num2}.")
elif num1 < num2:
    print(f"{num2} is greater than {num1}.")  
elif num1 == num2:
    print(f"{num1} is equal to {num2}.")
else:
    print("num1 and num2 are not equal.")
```

**Question 13:** Write a Python program to demonstrate logical operators: and, or, not.

```python
# Your code here
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

if num1 == 0 and num2 == 0:
    print("Both numbers are zero.")
elif num1 == 0 or num2 == 0:
    print(f"One of the numbers is zero: {num1} and {num2}.")
else:
    print("num1 and num2 are not equal.")
```

**Question 16:** Write a Python program to find the sum of the first n natural numbers.

```python
# Your code here
num = int(input("Enter a number: "))

if num > 0:
    sum_of_natural_numbers = num * (num + 1) // 2
    print(f"The sum of the first {num} natural numbers is: {sum_of_natural_numbers}")
else:
    print("Please enter a positive integer.")
```

**Question 17:** Write a Python program to check if a year is a leap year.

```python
# Your code here
year = int(input("Enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
    
```

**Question 19:** Write a Python program to check if a string is a palindrome.

```python
# Your code here
string = input("Enter a string: ")
if string == string[::-1]:
    print(f'"{string}" is a palindrome.')
else:
    print(f'"{string}" is not a palindrome.')
```
