```python
#Declaring and Assigning Variables

age = 32
height = 5.9
name = "Yashwanth"
is_student = False

print("Age:", age)
print("Height:", height)
print("Name:", name)
print("Is Student:", is_student)
```

```python
#Data Types

var = 10
name = "John"
float_num = 3.14
a = 1000
b = 2000
c = a == b

print("Integer:", type(var))
print("String:", type(name))
print("Float:", type(float_num))
print("Boolean:", type(c))
```

```python
result = "Hello" + str(10)
print(result)
```

```python
#Arithmetic Operations

a = 10
b = 5

c = a + b
print("Sum:", c)

d = a - b
print("Difference:", d)

e = a * b
print("Product:", e)

f = a / b
print("Quotient:", f)

g = a % b
print("Remainder:", g)

h = a ** b
print("Exponentiation:", h)
```

```python
#Comparison Operations

x = 10
y = 5

print("Equal:", x == y)
print("Not Equal:", x != y)
print("Greater Than:", x > y)
print("Less Than:", x < y)
print("Greater Than or Equal To:", x >= y)
print("Less Than or Equal To:", x <= y)
```

```python
#Logical Operations

p = 1
q = 0

print("Logical 1 AND 0:", p and q)
print("Logical 0 AND 1:", q and p)
print("Logical 1 OR 0:", p or q)
print("Logical 0 OR 1:", q or p)
print("Logical NOT 1:", not p)
print("Logical NOT 0:", not q)
```
